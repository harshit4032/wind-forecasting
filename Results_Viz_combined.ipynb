```python
# Fill missing model names
aug['Model'] = aug['Model'].fillna(method='ffill')
shuffle['Model'] = shuffle['Model'].fillna(method='ffill')

# Outer join to keep all models (e.g., RF and XGB)
merged = pd.merge(aug, shuffle, on=['Model', 'Metric'], how='left', suffixes=('_aug', '_shuffle'))

# Drop duplicate '(shuffle)' columns
cols_to_drop = [col for col in merged.columns if '(shuffle)' in col]
merged.drop(columns=cols_to_drop, inplace=True)

# Rename columns to standardized format
def rename_columns(df):
    col_map = {}
    for col in df.columns:
        if '(Org)' in col:
            col_map[col] = col.replace('(Org)', '').strip() + ' [Org (Shuffle)]'
        elif '(Aug)' in col:
            col_map[col] = col.replace('(Aug)', '').strip() + ' [Aug (Shuffle)]'
        elif '(no_shuffle)' in col:
            col_map[col] = col.replace('(no_shuffle)', '').strip() + ' [Org (No Shuffle)]'
    return df.rename(columns=col_map)

merged = rename_columns(merged)

# Now reorder columns: Model, Metric, and grouped location-wise
# Step 1: Extract the location names
location_prefixes = ['Location_1', 'Location_2', 'Location_3', 'Location_4', 'All Location Combined']
suffix_order = ['[Org (Shuffle)]', '[Aug (Shuffle)]', '[Org (No Shuffle)]']

# Step 2: Build ordered list of columns
ordered_cols = ['Model', 'Metric']
for loc in location_prefixes:
    for suffix in suffix_order:
        col = f"{loc} {suffix}"
        if col in merged.columns:
            ordered_cols.append(col)

# Step 3: Reorder the DataFrame
merged = merged[ordered_cols]
```

```python
import numpy as np

def to_list_mean_std(val):
    if isinstance(val, str) and '±' in val:
        try:
            mean_str, std_str = [x.strip() for x in val.split('±')]
            return [float(mean_str), float(std_str)]
        except:
            return [np.nan, np.nan]
    return [np.nan, np.nan]

# Make a copy to avoid overwriting original if needed
structured_results_df_cleaned = merged.copy()

# Apply transformation to all columns except 'Model' and 'Metric'
cols_to_process = merged.columns[2:]

for col in cols_to_process:
    structured_results_df_cleaned[col] = merged[col].apply(to_list_mean_std)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.colors as mcolors

# Constants
locations = ['Location_1', 'Location_2', 'Location_3', 'Location_4']
variants = ['Org (Shuffle)', 'Aug (Shuffle)', 'Org (No Shuffle)']
metrics = ['MAE', 'RMSE', 'R2']
variant_colors = {'Org (Shuffle)': 0.5, 'Aug (Shuffle)': 1.0, 'Org (No Shuffle)': 0.2}
variant_hatch = {'Org (Shuffle)': '', 'Aug (Shuffle)': '///', 'Org (No Shuffle)': '\\\\'}
df = structured_results_df_cleaned.copy()

# Models
base_models = sorted(df['Model'].unique())

# Color map: Model + Variant → RGBA color
base_palette = list(mcolors.TABLEAU_COLORS.values())
color_map = {}
for i, model in enumerate(base_models):
    base_color = base_palette[i % len(base_palette)]
    for variant in variants:
        color_map[f"{model} ({variant})"] = mcolors.to_rgba(base_color, alpha=variant_colors[variant])

# Plot for each metric
for metric in metrics:
    fig, ax = plt.subplots(figsize=(20, 6))
    x = np.arange(len(locations))  # 0, 1, 2, 3
    bar_width = 0.03
    all_model_variants = [f"{model} ({variant})" for model in base_models for variant in variants]
    total_bars = len(all_model_variants)
    total_width = total_bars * bar_width
    offset_center = (total_width - bar_width) / 2

    plotted_labels = set()

    for idx, model_variant in enumerate(all_model_variants):
        if ' (' not in model_variant:
            continue  # skip malformed
        model_name = model_variant.split(' (')[0]
        variant = model_variant[len(model_name)+2:-1]  # Extract inside the brackets

        df_row = df[(df['Model'] == model_name) & (df['Metric'] == metric)]
        if df_row.empty:
            continue

        means, stds = [], []
        for loc in locations:
            col = f"{loc} [{variant}]"
            val = df_row[col].values[0] if col in df_row else [np.nan, np.nan]
            means.append(val[0] if isinstance(val, list) else np.nan)
            stds.append(val[1] if isinstance(val, list) else np.nan)

        positions = x - offset_center + idx * bar_width
        label = model_variant if model_variant not in plotted_labels else None

        ax.bar(positions, means, yerr=stds, width=bar_width,
               color=color_map[model_variant],
               hatch=variant_hatch.get(variant, ''),
               label=label, capsize=3)

        if label:
            plotted_labels.add(model_variant)

    # Formatting
    title = f"{metric} Across Locations"
    if metric != "R2":
        title += " (Log Scale)"
        ax.set_yscale("log")

    ax.set_title(title, fontsize=16)
    ax.set_xlabel("Location", fontsize=14)
    ax.set_ylabel(metric, fontsize=14)
    ax.set_xticks(x)
    ax.set_xticklabels(locations, fontsize=12)
    ax.grid(True, axis="y", linestyle="--", alpha=0.6)
    ax.set_facecolor("white")
    fig.patch.set_facecolor("white")

    if metric == "R2":
        ax.axhline(y=0, color="gray", linestyle="--", linewidth=1)

    # Legend (only if labels exist)
    handles, labels = ax.get_legend_handles_labels()
    if labels:
        sorted_legend = sorted(zip(labels, handles), key=lambda x: x[0])
        labels, handles = zip(*sorted_legend)
        ax.legend(handles, labels, title="Model (Variant)", loc="center left",
                  bbox_to_anchor=(1.0, 0.5), ncol=1, fontsize=9)

    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_all_metrics_symlog_sorted(df, metrics, base_models,
                                    variants=['Org (Shuffle)', 'Aug (Shuffle)', 'Org (No Shuffle)'],
                                    sort_metric='MAE'):
    """
    Plots all metrics across models in one plot using symlog scale, sorted by a chosen strong metric (e.g., MAE).

    Parameters:
    - df: structured_results_df_cleaned
    - metrics: List of metric names like ['MAE', 'RMSE', 'MAPE', 'SMAPE', 'R2'].
    - base_models: List of model names.
    - variants: ['Org (Shuffle)', 'Aug (Shuffle)', 'Org (No Shuffle)']
    - sort_metric: Metric used to sort models (e.g., 'MAE')
    """
    col_map = {
        'Org (Shuffle)': 'All Location Combined [Org (Shuffle)]',
        'Aug (Shuffle)': 'All Location Combined [Aug (Shuffle)]',
        'Org (No Shuffle)': 'All Location Combined [Org (No Shuffle)]'
    }

    # --- Step 1: Sort models based on sort_metric (Original only) ---
    sort_df = df[df['Metric'] == sort_metric].copy()
    model_scores = []

    for model in base_models:
        row = sort_df[sort_df['Model'] == model]
        if row.empty or col_map['Org (Shuffle)'] not in row:
            continue
        value = row[col_map['Org (Shuffle)']].values[0]
        mean = value[0] if isinstance(value, list) else np.nan
        model_scores.append((model, mean))

    # Sort models: lower value is better for error metrics
    sorted_models = [m[0] for m in sorted(model_scores, key=lambda x: x[1])]

    # --- Step 2: Plot all metrics with models in same order ---
    fig, ax = plt.subplots(figsize=(14, 7))

    for i, metric in enumerate(metrics):
        offset = i * 0.15  # For separating metric lines slightly

        for j, variant in enumerate(variants):
            col = col_map[variant]
            filtered = df[df['Metric'] == metric].copy()

            means, stds = [], []

            for model in sorted_models:
                row = filtered[filtered['Model'] == model]
                if row.empty or col not in row:
                    means.append(np.nan)
                    stds.append(0)
                    continue
                value = row[col].values[0]
                if isinstance(value, list) and len(value) == 2:
                    mean, std = value
                else:
                    mean, std = np.nan, 0
                means.append(mean)
                stds.append(std)

            x = np.arange(len(sorted_models)) + offset + j * 0.05
            ax.errorbar(x, means, yerr=stds, fmt='-o', capsize=4,
                        label=f'{metric} ({variant})')

    ax.set_xticks(np.arange(len(sorted_models)))
    ax.set_xticklabels(sorted_models, rotation=45)
    ax.set_yscale('symlog', linthresh=0.1)
    ax.set_title(f"All Metrics Across Models (Sorted by {sort_metric}, symlog scale)", fontsize=15)
    ax.set_xlabel("Models", fontsize=12)
    ax.set_ylabel("Metric Value", fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.legend(ncol=2, fontsize=9)
    plt.tight_layout()
    plt.show()
```

```python
metrics = ['MAE', 'RMSE', 'MAPE', 'SMAPE', 'R2']
base_models = structured_results_df_cleaned['Model'].unique()

plot_all_metrics_symlog_sorted(
    df=structured_results_df_cleaned,
    metrics=metrics,
    base_models=base_models,
    variants=['Org (Shuffle)', 'Aug (Shuffle)', 'Org (No Shuffle)'],
    sort_metric='MAE'
)
```

```python
def merge_monthly_aug_shuffle(monthly_aug_df, monthly_shuffle_df):
    # Step 1: Rename Augmented and Org columns in monthly_aug_df
    aug_renamed = monthly_aug_df.rename(columns={
        col: col.replace('_aug_', '_Aug(shuffle)_') for col in monthly_aug_df.columns if '_aug_' in col
    }).rename(columns={
        col: col.replace('_org_', '_Org(shuffle)_') for col in monthly_aug_df.columns if '_org_' in col
    })

    # Step 2: Rename NoShuffle columns in monthly_shuffle_df
    shuffle_renamed = monthly_shuffle_df.rename(columns={
        col: col.replace('_no_shuffle_', '_Org(noshuffle)_') for col in monthly_shuffle_df.columns if '_no_shuffle_' in col
    })

    # Step 3: Select only renamed columns
    aug_cols = [col for col in aug_renamed.columns if '_Aug(shuffle)_' in col or '_Org(shuffle)_' in col]
    shuffle_cols = [col for col in shuffle_renamed.columns if '_Org(noshuffle)_' in col]

    # Step 4: Merge on 'month' and 'Model'
    merged_df = pd.merge(
        aug_renamed[['month', 'Model'] + aug_cols],
        shuffle_renamed[['month', 'Model'] + shuffle_cols],
        on=['month', 'Model'],
        how='outer'
    )

    return merged_df
```

```python
import matplotlib.pyplot as plt
import numpy as np

def lineplot_monthly_metric_all_versions(merged_df, metric="mae"):
    models = merged_df['Model'].unique()
    months = sorted(merged_df['month'].unique())

    plt.figure(figsize=(16, 8))

    for model in models:
        model_df = merged_df[merged_df['Model'] == model].sort_values('month')

        # Plot Original
        if f"{metric}_Org(shuffle)_mean" in model_df.columns:
            plt.errorbar(
                model_df['month'],
                model_df[f"{metric}_Org(shuffle)_mean"],
                yerr=model_df[f"{metric}_Org(shuffle)_std"],
                label=f'{model} - Org (shuffle)',
                linestyle='-',
                marker='o',
                capsize=3
            )

        # Plot Augmented
        if f"{metric}_Aug(shuffle)_mean" in model_df.columns:
            plt.errorbar(
                model_df['month'],
                model_df[f"{metric}_Aug(shuffle)_mean"],
                yerr=model_df[f"{metric}_Aug(shuffle)_std"],
                label=f'{model} - Aug (shuffle)',
                linestyle='--',
                marker='x',
                capsize=3
            )

        # Plot No Shuffle
        if f"{metric}_Org(noshuffle)_mean" in model_df.columns:
            plt.errorbar(
                model_df['month'],
                model_df[f"{metric}_Org(noshuffle)_mean"],
                yerr=model_df[f"{metric}_Org(noshuffle)_std"],
                label=f'{model} - Org (noshuffle)',
                linestyle=':',
                marker='s',
                capsize=3
            )

        # Plot Shuffled
        if f"{metric}_Shuffled_mean" in model_df.columns:
            plt.errorbar(
                model_df['month'],
                model_df[f"{metric}_Shuffled_mean"],
                yerr=model_df[f"{metric}_Shuffled_std"],
                label=f'{model} - Shuffled',
                linestyle='-.',
                marker='^',
                capsize=3
            )

    # Optional: Set y-ticks manually depending on metric
    y_min, y_max = plt.ylim()
    if metric in ["mae", "rmse", "mape", "smape"]:
        step = np.round((y_max - y_min) / 8, 2)
        yticks = np.arange(0, y_max + step, step)
        plt.yticks(yticks)
    elif metric == "r2":
        plt.yticks(np.linspace(-1, 1, 11))

    plt.title(f"Monthly {metric.upper()} Comparison Across Versions", fontsize=16)
    plt.xlabel("Month", fontsize=12)
    plt.ylabel(metric.upper(), fontsize=12)
    plt.xticks(rotation=45)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.legend(loc='upper right', fontsize=10, frameon=True, facecolor='white', framealpha=0.8)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def lineplot_monthly_metric_faceted_symlog(merged_df, metric="mae"):
    models = sorted(merged_df['Model'].unique())
    months = sorted(merged_df['month'].unique())

    n_models = len(models)
    n_cols = 2
    n_rows = int(np.ceil(n_models / n_cols))

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows), sharex=True)
    axes = axes.flatten()

    # Define styles for each version
    version_styles = {
    'Aug(shuffle)': {'label': 'Aug (shuffle)', 'linestyle': '--', 'marker': 'o', 'color': '#2ca02c'},
    'Org(shuffle)': {'label': 'Org (shuffle)', 'linestyle': '-.', 'marker': '^', 'color': '#d62728'},
    'Org(noshuffle)': {'label': 'Org (noshuffle)', 'linestyle': '-', 'marker': 's', 'color': '#1f77b4'},
        }


    used_labels = set()  # Track used labels for legend

    for i, model in enumerate(models):
        ax = axes[i]
        model_df = merged_df[merged_df['Model'] == model].sort_values('month')

        plotted = False  # Track if anything was plotted

        for version, style in version_styles.items():
            mean_col = f"{metric}_{version}_mean"
            std_col = f"{metric}_{version}_std"

            if (
                mean_col in model_df.columns and
                std_col in model_df.columns and
                not model_df[mean_col].isna().all()
            ):
                ax.errorbar(
                    model_df['month'],
                    model_df[mean_col],
                    yerr=model_df[std_col],
                    label=style['label'] if style['label'] not in used_labels else "",
                    linestyle=style['linestyle'],
                    marker=style['marker'],
                    color=style['color'],
                    capsize=3,
                    elinewidth=1
                )
                used_labels.add(style['label'])
                plotted = True
            else:
                print(f"⚠️ Skipping {model} - {version}: No valid data")

        if plotted:
            ax.set_yscale('symlog')
            ax.set_title(model, fontsize=12)
            ax.set_ylabel(metric.upper())
            ax.set_xticks(months)
            ax.set_xticklabels(months, rotation=45, fontsize=9)
            ax.grid(True, linestyle='--', alpha=0.5)
        else:
            ax.set_visible(False)

    # Remove unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    fig.suptitle(f'Monthly {metric.upper()} Comparison per Model (Symlog Scale)', fontsize=18)
    fig.tight_layout(rect=[0, 0.03, 1, 0.95])

    # Custom legend based on used labels
    handles = []
    labels = []
    for version, style in version_styles.items():
        if style['label'] in used_labels:
            line = plt.Line2D([0], [0], color=style['color'], linestyle=style['linestyle'], marker=style['marker'])
            handles.append(line)
            labels.append(style['label'])

    fig.legend(
        handles, labels,
        loc='upper center',
        ncol=len(handles),
        frameon=True,
        fontsize=12
    )

    plt.show()

```

```python
lineplot_monthly_metric_faceted_symlog(merged_monthly_df, metric='mae')
```

```python
import matplotlib.pyplot as plt
import numpy as np
import itertools

def lineplot_monthly_metric_all_versions_logscale(merged_df, metric="mae"):
    models = merged_df['Model'].unique()
    months = sorted(merged_df['month'].unique())

    # Define a wide color palette (20 distinct colors)
    color_palette = plt.cm.tab20(np.linspace(0, 1, 20))
    color_cycle = itertools.cycle(color_palette)

    # Define line styles and markers for each version
    version_styles = {
        'Org(shuffle)': {'linestyle': '-', 'marker': 'o'},
        'Aug(shuffle)': {'linestyle': '--', 'marker': 'x'},
        'Org(noshuffle)': {'linestyle': ':', 'marker': 's'},
        'Shuffled': {'linestyle': '-.', 'marker': '^'}
    }

    plt.figure(figsize=(18, 10))

    for model in models:
        model_df = merged_df[merged_df['Model'] == model].sort_values('month')
        base_color = next(color_cycle)  # Assign one base color per model

        for i, (version, style) in enumerate(version_styles.items()):
            col_mean = f"{metric}_{version}_mean"
            col_std = f"{metric}_{version}_std"

            if col_mean in model_df.columns:
                # Slight variation in color for each version of the model
                color_variant = base_color * (0.85 + 0.05 * i)  # Brighten/darken
                color_variant = np.clip(color_variant, 0, 1)

                plt.errorbar(
                    model_df['month'],
                    model_df[col_mean],
                    yerr=model_df[col_std],
                    label=f'{model} - {version}',
                    linestyle=style['linestyle'],
                    marker=style['marker'],
                    capsize=3,
                    color=color_variant
                )

    plt.yscale('log')
    plt.title(f"Monthly {metric.upper()} Comparison Across Versions (Log Scale)", fontsize=16)
    plt.xlabel("Month", fontsize=13)
    plt.ylabel(metric.upper(), fontsize=13)
    plt.xticks(rotation=45)
    plt.grid(True, linestyle='--', alpha=0.5, which='both')
    plt.legend(loc='upper right', fontsize=10, frameon=True, facecolor='white', framealpha=0.8)
    plt.tight_layout()
    plt.show()
```
